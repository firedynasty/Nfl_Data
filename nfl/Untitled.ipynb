```python
import pandas as pd
nfl_data = pd.read_csv('2022_NFL_Season_Scores.csv')
```

```python
import pandas as pd


# Group by Pass Yds Player and Pass Yds Team, then count occurrences and sum passing yards
top_pass_yds_with_counts = nfl_data.groupby(['Pass Yds Player', 'Pass Yds Team'])['Pass Yds'].agg(['count', 'sum']).reset_index()

# Sort the results by count and then by sum of passing yards
top_pass_yds_with_counts = top_pass_yds_with_counts.sort_values(by=['count', 'sum'], ascending=[False, False])

# Display the result
top_pass_yds_with_counts
```

```python
# Group by Pass Yds Player and Pass Yds Team, then count occurrences and sum passing yards
top_pass_yds_with_counts = nfl_data.groupby(['Pass Yds Player', 'Pass Yds Team'])['Pass Yds'].agg(['count', 'sum']).reset_index()

# Sort the results by count and then by sum of passing yards
top_pass_yds_with_counts = top_pass_yds_with_counts.sort_values(by=['count', 'sum'], ascending=[False, False]).reset_index(drop=True)

# Display the result
top_pass_yds_with_counts.to_csv('top_quarterbackers_2022.csv', index=False)
```
